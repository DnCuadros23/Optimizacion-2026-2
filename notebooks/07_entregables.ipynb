## Tablas finales en LaTeX (listas para pegar en Overleaf)

```python
for nombre in ("berlin52_resumen_calibrado.tex", "eil51_resumen.tex", "calibracion_resumen.tex", "optimo_subconjuntos.tex"):
    ruta = RAIZ / "results" / "tables" / "latex" / nombre
    print(f"--- {nombre} ---")
    print(ruta.read_text(encoding="utf-8"))
    print()
```
